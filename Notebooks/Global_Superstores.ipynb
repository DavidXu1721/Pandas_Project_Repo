```python
category_profit_summary = df.groupby('Category')['Profit'].agg(['sum', 'mean', 'count', 'min', 'max']).reset_index()
category_profit_summary.columns = ['Category', 'Total Profit', 'Average Profit', 'Number of Orders', 'Min Profit', 'Max Profit']
display(category_profit_summary)
```

```python
segments_sales_summary = df.groupby('Segment')['Sales'].agg(['sum', 'mean', 'count', 'min', 'max']).reset_index()
segments_sales_summary.columns = ['Segment', 'Total Sales', 'Average Sales', 'Number of Orders', 'Min Sales', 'Max Sales']
display(segments_sales_summary)
```
